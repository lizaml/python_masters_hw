# 1. Анализ типов

| Выражение | Тип | bool(...) | Почему |
|---|---|---|---|
|42 | int | True |все int, кроме 0, True|
|19.99 | float | True |все float, кроме 0, True|
|'' | str | False |str длинной 0 дает False|
|'Python' | str | True |str длинной > 0 дает True|
|[] | list | False |list длинной 0 дает False|
|[0] | list | True |list длинной > 0 дает True|
|None | NoneType | False |NoneType всегда дает False|
|0 | int | False |int 0 дает False|
|0.001 | float | True |float > 0 дает True|
|False | bool | False |bool False дает False|

```python
print("| Выражение | Тип | Truthy/Falsy | Почему |")
print("|---|---|---|---|")
for x in [42, 19.99, "", "Python", [], [0], None, 0, 0.001, False]:
    print(f'|{x!r} | {type(x).__name__} | {bool(x)} ||')
```

# 2. Операторы // и %

```python
def integer_analyse(n: int) -> None:
    print(f'Полных десятков: {n // 10}')
    print(f'Последняя цифра: {n % 10}')
    print(f'Четное: {n % 2 == 0}')
    print(f'Кратно 5: {n % 5 == 0}')

integer_analyse(int(input('Введите целое число: ')))
```

# 4. Работа со списками и enumerate

```python
countries = ["Germany", "France", "Italy", "Spain", "Portugal"]
```

1. выводит страны в формате

```python
for i, country in enumerate(countries, start=1):
    print(f"{i}. {country}")
```

2. создаёт новый список, содержащий только страны, длина названия которых больше 5 символов

```python
countries_more_than_5 = [country for country in countries if len(country) > 5]
print(countries_more_than_5)
```

3. подсчитывает общее количество букв во всех названиях стран

```python
total_letters = 0

for country in countries:
    total_letters += len(country)

print(f'Общее количество букв: {total_letters}')
```

# 5. Мини-отчёт в JSON

```python
import json

completed_tasks = int(input('Введите количество выполненных заданий: '))

summary = {
    "student": "Елизавета Мацак",
    "completed_tasks": completed_tasks,
    "score": completed_tasks * 20,
    "passed": completed_tasks * 20 >= 60
}
```

```python
with open('summary.json', 'w') as f:
    json.dump(summary, f, indent=2)
```

```python
with open('summary.json') as f:
    summary_opened = json.load(f)
```

```python
print('score:', summary_opened['score'])
print('passed:', summary_opened['passed'])
```
