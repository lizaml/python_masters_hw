```python
def classify_temperature(temp: float) -> str:
    if temp < 0:
      return 'freezing'
    elif temp <= 15:
      return 'cold'
    elif temp <= 25:
      return 'warm'
    elif temp > 25: 
      return 'hot'
    else:
      return 'неизвестно'
```

```python
classify_temperature(float('nan'))
```
