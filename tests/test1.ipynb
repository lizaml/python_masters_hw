# 1

Напишите функцию, которая возвращает оценку за курс:

- от 0 до 40 - оценка "Неуд"
 
- от 40 до 60 - оценка "Удовлетворительно"
 
- от 60 до 85 - оценка "Хорошо"
 
- от 85 до 100 - оценка "Отлично"

```python
def get_grade(score: int) -> str:
    if not isinstance(score, int):
      return 'Ошибка типа данных'
    
    if score < 0:
      return 'Ошибка, score < 0'
    elif score <= 40:
      return 'Неуд'
    elif score <= 60:
      return 'Удовлетворительно'
    elif score <= 85:
      return 'Хорошо'
    elif score <= 100:
      return 'Отлично'
    else:
      return 'Ошибка'
```

```python
get_grade(-1) 
```

# 2 

Из числа x создаётся последовательность длины n, в которой числа x и -x чередуются, начиная с x.

Например, при x = 3 и n = 5 последовательность такая: 3, -3, 3, -3, 3.

Функция возвращает сумму всех чисел последовательности.

```python
# Если в лоб с созданием последовательности
def alternating_sum(x: int, n: int) -> int:
    if not isinstance(x, int):
      return 'Ошибка'
    if not isinstance(n, int):
      return 'Ошибка'
    if n < 0:
      return 'Ошибка'

    total = 0
    for i in range(n):
      if i % 2 == 0:
        total += x
      else:
        total += -x

    return total
```

```python
# Вариант 2 – без суммирования
def alternating_sum_v2(x: int, n: int) -> int:
    if not isinstance(x, int):
      return 'Ошибка'
    if not isinstance(n, int):
      return 'Ошибка'
    if n < 0:
      return 'Ошибка'
    
    if n % 2 == 0:
      return 0
    else:
      return x
```

```python
print(alternating_sum(444, 3))
print(alternating_sum_v2(444, 3))
```

# 3

Функция, которая возвращает количество четных цифр в числе
- 123 - 1 четная цифра 2
- 77666 - 3 четные цифры (3 шестерки)
- 880888 - 6 четных цифр (восмерки и 0)

Гарантируется что входное число неотрицательное


```python
def count_even_numbers(number: int) -> int:
    # Проверяем тип данных
    if not isinstance(number, int):
      return 'Ошибка'

    count = 0
    for digit in str(number):
      if int(digit) % 2 == 0:
        count += 1

    return count
```
